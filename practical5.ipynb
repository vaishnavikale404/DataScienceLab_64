```python
df = pd.read_csv("student_performance.csv", nrows=10000)
df.head()
```

Dataset Information

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nDataset Description:")
print(df.describe())
```

Data Filtering

```python
high_score = df[df['total_score'] > 90]

print("\nStudents with score > 90:")
print(high_score.head())
```

Create Category Column

```python
df['Score_Category'] = df['total_score'].apply(
    lambda x: 'High' if x > 85 else 'Medium' if x > 60 else 'Low'
)

print(df.head())
```

Grouping Data

```python
grouped_data = df.groupby('Score_Category')

print("\nGrouped Data:")
print(grouped_data.head())
```

Mean Score per Category

```python
print("\nMean Score per Category:")
print(df.groupby('Score_Category')['total_score'].mean())
```

Count Students per Category

```python
print("\nNumber of Students per Category:")
print(df.groupby('Score_Category')['total_score'].count())
```

Minimum Score

```python
print("\nMinimum Score:")
print(df['total_score'].min())
```

Maximum Score

```python
print("\nMaximum Score:")
print(df['total_score'].max())
```
